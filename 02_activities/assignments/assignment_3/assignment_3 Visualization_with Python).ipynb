#### The dataset I used it for this visualization : https://data.ontario.ca/dataset/wages-by-education-level/resource/4cc07c1b-62ed-4ece-a2a4-d05d0f45081c

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("assignment_3/Data/data.csv")
```

```python
clean_data = data.dropna(subset=['AGE GROUP', 'Men', 'Women'])

mean_wages = clean_data.groupby('AGE GROUP')[['Men', 'Women']].mean()

overall_mean = clean_data[['Men', 'Women']].mean()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Bar chart
mean_wages.plot(kind='bar', ax=ax1, width=0.7, color=['blue', 'pink'])
ax1.set_title('Mean Wages by Age Group and Gender')
ax1.set_xlabel('Age Group')
ax1.set_ylabel('Mean Wage Rate')

ax1.legend(title='Gender', bbox_to_anchor=(1.05, 1), loc='upper left')

# Pie chart
ax2.pie(overall_mean, labels=['Men', 'Women'], autopct='%1.1f%%', colors=['blue', 'pink'])
ax2.set_title('Overall Mean Wage Distribution by Gender')

plt.show()
```
