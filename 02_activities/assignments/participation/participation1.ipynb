```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
```

```python
# Set the random seed for reproducibility — ensures the same random numbers are generated each time the code is run
np.random.seed(613)

# Create an array of integers from 0 to 49 (used as x-axis values)
x = np.arange(50)

# Generate 50 random integers between 0 and 100 (used as y-axis values)
y = np.random.randint(0, 100, 50)

# Create a figure and axes object with a specific size of 5 inches wide by 3 inches tall
fig, ax = plt.subplots(figsize=(5, 3))

# Plot a histogram of the y values — shows the frequency distribution of the random integers
ax.hist(y)

# Other plot types that could be used instead (currently commented out):
# ax.scatter(x, y)     # Scatter plot
# ax.bar(x, y)         # Vertical bar chart
# ax.barh(x, y)        # Horizontal bar chart
# ax.plot(x, y)        # Line plot
```

```python
# Define custom font settings for title and axis labels
font1 = {"family": "sans-serif", "color": "blue", "size": 20}   # For the plot title
font2 = {"family": "monospace", "color": "green", "size": 14}   # For the x and y axis labels

# Create a figure and axes object with a specified size (5 inches by 3 inches)
fig, ax = plt.subplots(figsize=(5, 3))

# Plot the data as a line plot
ax.plot(x, y)

# Set the title of the plot, using font1 and aligning it to the right
ax.set_title("Total growth over time", fontdict=font1, loc="right")

# Set the label for the y-axis using font2
ax.set_ylabel("Total growth", fontdict=font2)

# Set the label for the x-axis using font2
ax.set_xlabel("Years since Start", fontdict=font2)
```

```python
# Create a figure and axes object with a specified size (5 inches wide by 3 inches tall)
fig, ax = plt.subplots(figsize=(5, 3))

# Create a scatter plot using x and y values
ax.scatter(
    x,           # x-axis values
    y,           # y-axis values
    marker="*",  # Use a star-shaped marker for each point
    color="indigo"  # Set the color of the markers to indigo
)
```

```python
# Create a figure and axes object with a specified size (5 inches by 3 inches)
fig, ax = plt.subplots(figsize=(5, 3))

# Plot the data as a line chart with customized markers and styling
ax.plot(
    x,                  # x-axis values
    y,                  # y-axis values
    marker="*",         # Use a star marker at each data point
    color="#7425B9",    # Line color (a shade of purple)
    linestyle="--",     # Dashed line style
    linewidth=2,        # Thickness of the line
    markersize=12,      # Size of the marker
    markeredgecolor="#FA9359",  # Color of the marker edge (orange)
    markerfacecolor="#000000"   # Fill color of the marker (black)
)

# Add horizontal gridlines only (along the y-axis), in blue with line width 2
ax.grid(axis="y", color="blue", linewidth=2)
```
